# Week 2: grid-search results
Compare the 24 unique architecture, width, loss and learning-rate settings. These tables use the saved validation results; no training or test-set selection occurs here.


```python
from cs163_rfi import paths
import pandas as pd
from IPython.display import display
```

```python
ranked = pd.read_csv(paths.RESULTS_DIR / "reference/grid_24.csv")
ranked = ranked.sort_values("pixel_ap", ascending=False).reset_index(drop=True)
display(ranked[["architecture", "base_channels", "loss", "learning_rate", "pixel_ap", "recall", "precision", "fpr"]])
```

## Marginal means
Each unique configuration has equal weight. Undefined precision stays missing. AP is exact for 65,536 quantized scores; it approximates AP of the original floating-point probabilities.


```python
metrics = ["pixel_ap", "recall", "precision", "fpr"]
for factor in ["architecture", "base_channels", "loss", "learning_rate"]:
    print(factor)
    display(ranked.groupby(factor)[metrics].mean())
```
